```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
def draw(src, detected_circles):

    image = src.copy()
    circles_count = 0

    if detected_circles is not None:
        circles = np.around(detected_circles).reshape(-1, 3)

        for points in circles:
            a, b, r = int(points[0]), int(points[1]), int(points[2])
            
            cv.circle(image, (a, b), r, (0, 255, 0), 2)
            cv.circle(image, (a, b), 1, (255, 0, 0), 3)
            
            circles_count += 1

    print("Total circles:", circles_count)
    plt.imshow(image, cmap='gray')
    plt.axis('off')
    plt.show()
    
    return 
```

```python
src = cv.imread("logs-1.jpg")
src = cv.cvtColor(src, cv.COLOR_BGR2RGB)
gray = cv.cvtColor(src, cv.COLOR_RGB2GRAY)
gray = cv.GaussianBlur(gray, (13, 13), 2) 
```

```python
detected_circles = cv.HoughCircles(
                   gray,
                   cv.HOUGH_GRADIENT_ALT,
                   dp=1,
                   minDist=25,     
                   param1=320,           
                   param2=0.79,          
                   minRadius=20,        
                   maxRadius=75        
)
```

```python
draw(src, detected_circles)
```
